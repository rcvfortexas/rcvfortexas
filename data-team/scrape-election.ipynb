```python
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.support.ui import Select
```

```python
import time

def wait_for_load_find_element_and_click(driver: webdriver, by, value, timeout=5):
    start = time.time()
    while time.time() < start + timeout:
        try:
            elem = driver.find_element(by, value)
            elem.click()
            print("clicked!")
            return
        except Exception:
            time.sleep(1)
    raise TimeoutError
```

```python
base_url = "https://results.texas-election.com/races"
election_title = "2024 MARCH 5TH REPUBLICAN PRIMARY"
election_category = "DISTRICT OFFICES"
```

```python
driver = webdriver.Firefox()
driver.implicitly_wait(5)
```

```python
driver.get(base_url)

wait_for_load_find_element_and_click(driver, By.CLASS_NAME, "electionVal")
wait_for_load_find_element_and_click(driver, By.CSS_SELECTOR, "mat-select[placeholder='Select Election']")
wait_for_load_find_element_and_click(driver, By.XPATH, f"//*[text()='{election_title}']")
wait_for_load_find_element_and_click(driver, By.XPATH, f"//span[text()='Update Election ']")
print("updating election...")
time.sleep(2)
wait_for_load_find_element_and_click(driver, By.XPATH, f"//div[contains(text(), '{election_category}')]")
```
